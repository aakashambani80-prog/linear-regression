```python
bf=pd.read_csv(r'C:\Users\bhush\Downloads\play_tennis.csv')
```

```python
l2=LabelEncoder()
for i in bf.columns:
    bf[i]=l2.fit_transform(bf[i])
```

```python
bf.to_csv('tenis-cleaned.csv')
```

```python
X=bf.iloc[:,:4]
y=bf['Play']
```

```python
X_train, X_test, y_train, y_test =train_test_split(X,y,test_size=0.2)
```

```python
dc=DecisionTreeClassifier(criterion='entropy')
```

```python
dc.fit(X_train,y_train)
```

```python
# y_pr1=dc.predict(X_test)
```

```python
plt.figure(figsize=(12, 8))
plot_tree(
    dc,
    feature_names=X.columns,  # Column names
    class_names=['0','1'],     # Target labels
    filled=True,                       # Color nodes by class
    rounded=True,                      # Rounded node boxes
    fontsize=10
)
plt.show()
```
